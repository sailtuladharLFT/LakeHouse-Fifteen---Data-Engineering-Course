## Handling Missing Values

1. Dropping Rows
2. Various parameters in dropping functionalities
3. handling missing values by using Mean. median and mode


```python
incomplete_df = spark.read.csv("test2.csv" , header=True , inferSchema = True)
```

```python
incomplete_df.show()
```

```python
incomplete_df.na.drop(how = 'any' , thresh=2).show()
```

```python
incomplete_df.na.drop(subset=['Experience']).show()
incomplete_df.na.drop(subset=['Name']).show()
```

```python
## Filling the Missing Values  

incomplete_df.na.fill("NaN").show() ## Compatible with string type columns only
incomplete_df.na.fill(0).show() ## Compatible with int type columns only

```

```python
incomplete_df.na.fill({
    'Name': "Unknown Name",
    "age" : 0,
    "Experience" : 0,
    "salary": 0
}).show()
```

```python
## Fill using mean , median and mode

from pyspark.ml.feature import Imputer

imputer_mean= Imputer(
    inputCols=['age','Experience'],
    outputCols=[ 'age' , 'Experience'],
    strategy='mean'
)

model_mean = imputer_mean.fit(incomplete_df)

result_mean = model_mean.transform(incomplete_df)

result_mean.show()
```

```python
## Fill using mean , median and mode

from pyspark.ml.feature import Imputer

imputer_median= Imputer(
    inputCols=['age','Experience'],
    outputCols=[ 'age' , 'Experience'],
    strategy='median'
)

model_median = imputer_median.fit(incomplete_df)

result_median = model_median.transform(incomplete_df)

result_median.show()
```

```python
## Fill using mean , median and mode

from pyspark.ml.feature import Imputer

imputer_mode= Imputer(
    inputCols=['age','Experience'],
    outputCols=[ 'age' , 'Experience'],
    strategy='mode'
)

model_mode = imputer_mode.fit(incomplete_df)

result_mode = model_mode.transform(incomplete_df)

result_mode.show()
```

### 1. Array Type
- They are simple lists type
- They can be nested within the columns

```python
# Arrays

data = [
    (1, "Ram", ["Python", "SQL", "Spark"]),
    (2, "Sam", ["Java", "SQL"]),
    (3, "Hari", ["Python", "Airflow"])
]

df = spark.createDataFrame(
    data,
    ["id", "name", "skills"]
)
```

```python
df.show(truncate=False)
```

```python
df.printSchema()
```

```python
#Accessing elements of an array

df.select(
    "name",
    df.skills[1].alias("Second Skill")
).show()

```

```python
# Explode: Converts each array element into a single row

from pyspark.sql.functions import explode

exploded_data = df.select(
    "id",
    "name",
    explode("skills").alias("skill")
)

exploded_data.show()
```

### Map Type: It is structure that represents a key value pair format

#### Eg: {"Python": 90, "SQL": 85}

```python


data = [
    (1, "Ram", {"Python": 90, "SQL": 85}),
    (2, "Sam", {"Java": 88, "SQL": 92}),
    (3, "Hari", {"Python": 95, "Airflow": 80})
]

df_map = spark.createDataFrame(
    data,
    ["id", "name", "scores"]
)

type(df_map.scores) # We can explode this column

df_map.schema["scores"].dataType
```

```python
df_map.select(
    "name",
    "scores.Python"
).fillna(0).show()
```

```python
df_map.show(truncate=False)

df_map.printSchema()

# Accessing Map Values

from pyspark.sql.functions import map_keys


old_df = df_map.select(
    "name",
    df_map.scores.Python.alias("Python Score"),
    df_map.scores.SQL.alias("SQL Score"),
    df_map.scores.Airflow.alias("Airflow Score")
).na.fill(
    {
        "Python Score" : 0,
        "SQL Score" : 0,
        "Airflow Score" : 0

    }
).show()

new_df = df_map.select(
    "name",
    df_map.scores.Python.alias("Python Score"),
    df_map.scores.SQL.alias("SQL Score"),
    df_map.scores.Airflow.alias("Airflow Score")
)

imputer = Imputer(
    inputCols = ['Python Score' , 'Airflow Score' , 'SQL Score'],
    outputCols=  ['Python Score' , 'Airflow Score' , 'SQL Score'],
    strategy=  "mean"
)

model_mean = imputer.fit(new_df)

result_df = model_mean.transform(new_df)

result_df.show()

```

```python
# Accessing the map keys 

df_1 = df_map.select(
    "name",
    map_keys("scores").alias("subjects")
)

df_1.show()

df_1.select(
    "name",
    explode('subjects').alias("subject")
).show()
```

```python
#Exploding entire maps

df_map.select(
    "name",
    explode("scores").alias("subject", "score")
).show()
```

### 3. Struct Type

- Similar to as Map type where it contains key value pairing, but it has a strict key value pariring
- The keys are pre-defined where as in map the keys are not predefined

```python
data = [
    (1, ("Ram", 25, "Kathmandu")),
    (2, ("Sam", 30, "Pokhara")),
    (3, ("Hari", 22, "Lalitpur"))
]

from pyspark.sql.types import StructType , StructField

schema = StructType([
    StructField("id" , IntegerType() , False),

    StructField('person' ,
        StructType([
                StructField("name" , StringType() , True),
                StructField("age", IntegerType(), True),
                StructField("city", StringType(), True)
            ]), True)
        
])

df_struct = spark.createDataFrame(data , schema)
```

```python
df_struct.show()

df_struct.printSchema()
```

```python
# Accessing the nested structs

df_struct.select(
    "person.name",
    "person.age",
    "person.city"
).show()

type(df_struct.person)
df_struct.schema["person"].dataType

```

#### Note: Explode only works for ArrayType and MapType not for StructType

### 3. JSON Type

- str containing JSON
'{"name":"Ram","age":25}'




```python
data = [
    (1, '{"name":"Ram","age":25,"city":"Kathmandu"}'),
    (2, '{"name":"Sam","age":30,"city":"Pokhara"}')
]

df_json = spark.createDataFrame(
    data,
    ["id", "person_json"]
)
```

```python
df_json.printSchema()
```

```python
#from_json: it is used to convert json data that is stored as string within a column as a structured data as a key value pair similar to nested struct

person_schema = StructType([
    StructField("name" , StringType() , True),
    StructField("age" , IntegerType() , True),
    StructField("city" , StringType() , True)

])
```

```python
from pyspark.sql.functions import from_json

parsed_df = df_json.withColumn(
    "person",
    from_json("person_json" , person_schema)
)
```

```python
parsed_df.printSchema()

parsed_df.show()
```

# Hands on Lab Practice

```python
data = [
    (1, '''
    {
        "name": "Laptop",
        "categories": ["electronics", "computers"],
        "prices": {
            "USD": 1000,
            "NPR": 140000
        }
    }
    '''),
    
    (2, '''
    {
        "name": "Phone",
        "categories": ["electronics", "mobile"],
        "prices": {
            "USD": 700,
            "NPR": 98000
        }
    }
    ''')
]

df_json = spark.createDataFrame(data , ['id' , 'product_json'])
```

```python
df_json.show(truncate=False)

df_json.printSchema()
```

```python
product_schema = StructType([
    StructField("name", StringType() , True),
    StructField("categories", ArrayType(StringType() , True) , True),
    StructField("prices", MapType(StringType(), IntegerType()) , True),
])
```

```python
parsed_df = df_json.withColumn(
    "details",
    from_json("product_json" , product_schema)
).drop("product_json")
```

```python
parsed_df.show(truncate=False)

parsed_df.schema['details'].dataType
```

```python
final_df = parsed_df.select(
    "id",
    "details.name",
    explode("details.categories").alias("category"),
    explode("details.prices").alias("currency" , "prices")
)
```

```python
final_df.show()
```

## Group By and Aggregate functions

- Group By and aggregate function works together
- For that we need to implement group by first then aggregate

```python
test_3_df = spark.read.csv("test3.csv" , header=True , inferSchema=True)
```

```python
test_3_df.show()
```

```python
# Highest Salary

test_3_df.groupBy("Name").sum('salary').show()

# Department monthly salary expense per person on AVG

test_3_df.groupBy("Departments").avg().show()

# Highest Earners in each depart

test_3_df.groupBy("Departments").max().show()
```
